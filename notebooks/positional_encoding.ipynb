# Positional Encoding from Scratch

Sinusoidal + learned PE in NumPy for `ai-learn-02-positional-encoding`.

$$PE(pos,2i)=\sin(pos/10000^{2i/d}),\quad PE(pos,2i+1)=\cos(\cdots)$$

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
sys.path.insert(0, str(ROOT if (ROOT/'positional_encoding.py').exists() else ROOT.parent))
from positional_encoding import *
print('ok')
```

```python
T,D=64,32
pe=sinusoidal_positional_encoding(T,D)
print(pe.shape, pe.min(), pe.max())
assert all(check_sinusoidal_properties(pe).values())
```

```python
plt.figure(figsize=(10,3)); plt.imshow(pe.T,aspect='auto',cmap='RdBu',vmin=-1,vmax=1); plt.title('PE heatmap'); plt.xlabel('pos'); plt.ylabel('dim'); plt.colorbar(); plt.show()
```

```python
sim=position_cosine_similarity(pe)
plt.figure(figsize=(4,3)); plt.imshow(sim,cmap='viridis',vmin=-1,vmax=1); plt.title('PE cosine sim'); plt.colorbar(); plt.show()
print('adj', np.mean([sim[i,i+1] for i in range(T-1)]), 'far', sim[0,-1])
```

```python
rng=np.random.default_rng(0); emb=rng.normal(0,0.1,(8,D)); seq=np.array([1,3,1,5])
tok=emb[seq]; with_pe=add_positional_encoding(tok, sinusoidal_positional_encoding(len(seq),D))
print('same token before', np.allclose(tok[0],tok[2]), 'after differ', not np.allclose(with_pe[0],with_pe[2]))
```
